# Lecture 3 - Customizing Plots

legends, annotations, ticks, styles

```python
import numpy as np
import matplotlib.pyplot as plt

# two y variables this time
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

## multiple lines on same axes

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1)
ax.plot(x, y2)
```

## legends
need label argument + call legend()

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc='lower right')
```

```python
# more legend options
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc='lower right',
          frameon=True,
          fontsize=12,
          ncol=2,       # horizontal layout
          shadow=True)
```

```python
# legend outside plot - bbox_to_anchor
# (1,1) = top right corner of plot
# loc='upper left' means legend's upper left attaches there
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
```

## text annotations

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

# text at coordinates (10, 95)
ax.text(10, 95, "This value is important!")
```

```python
# text customization
# ha = horizontal alignment (left, center, right)
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!",
        ha='center', color='red', size=20)
```

## transform positioning
- transData = data coordinates
- transAxes = axes coords (0-1)
- transFigure = figure coords (0-1)

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData)
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure)
```

## arrow annotations
annotate() - points to specific location with arrow

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

# xy = point arrow goes to, xytext = where text sits
ax.annotate('This is important!',
            xy=(10, 95),
            xytext=(20, 94),
            arrowprops=dict(facecolor='black'))
```

```python
# different arrow styles
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.annotate('This is important!',
            xy=(10, 95), xytext=(20, 94),
            arrowprops=dict(arrowstyle="wedge", color="hotpink"))
```

## tick marks
- NullLocator = remove ticks entirely
- NullFormatter = remove labels but keep ticks

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1)
ax.scatter(x, y2)

ax.yaxis.set_major_locator(plt.NullLocator())    # no y ticks
ax.xaxis.set_major_formatter(plt.NullFormatter())  # no x labels
```

```python
# MaxNLocator - limit number of ticks
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1)
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
# MultipleLocator - set tick interval
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
# rotate tick labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1)
plt.xticks(rotation=45, ha='right')
```
