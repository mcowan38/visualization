# Lecture 1 - Intro to Matplotlib

basics of making plots in python

```python
# imports
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests
```

```python
# sample data - seed for reproducibility
np.random.seed(613)
x = np.arange(50)  # 0 to 49
y = np.random.randint(0, 100, 50)  # 50 random ints
```

## fig vs ax
- fig = the whole canvas/window
- ax = the actual plot area (can have multiple)

```python
# scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y)
```

```python
# bar plot - same setup, different method
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x, y)
```

```python
# line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
```

```python
# histogram - only needs y (shows distribution)
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

## titles and labels

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

## font customization with fontdict

```python
# define font styles as dictionaries
font1 = {'family': 'sans-serif', 'color': 'blue', 'size': 20}
font2 = {'family': 'monospace', 'color': 'green', 'size': 14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
```

## customizing markers and colors
note: matplotlib uses american spelling (color not colour)

```python
# star markers, indigo color
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y, marker='*', color='indigo')
```

```python
# line style and width
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', color='indigo', linestyle='--', linewidth=2)
```

```python
# can use hex codes for colors
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', color='#7425b9', linestyle='--', linewidth=2)
```

```python
# full marker customization - size, edge color, face color
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*',
        markersize=12,
        color='#7425b9',
        linestyle='--',
        linewidth=2,
        markeredgecolor='#fa9359',
        markerfacecolor='#fa9359')
```

## grid lines

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.grid(axis='y')  # only y-axis grid
```

```python
# customize grid appearance
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.grid(axis='y', color='blue', linewidth=2, linestyle='-.')
```
